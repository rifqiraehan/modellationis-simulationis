#### 4.4 Simulasi 24 Jam

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(10)
lam_per_hour = 30
hours = 24
arrivals = np.random.poisson(lam_per_hour, hours)

print("Total pelanggan :", arrivals.sum())
print("Rata-rata/jam   :", arrivals.mean())
print("Jam tersibuk    :", arrivals.argmax() + 1)
print("Jumlah tersibuk :", arrivals.max())

plt.plot(range(1, 25), arrivals, marker="o")
plt.xlabel("Jam")
plt.ylabel("Jumlah pelanggan")
plt.title("Simulasi Kedatangan Pelanggan 24 Jam")
plt.grid(True)
plt.show()
```

### Analisis Singkat

Simulasi menunjukkan total 740 pelanggan dalam 24 jam dengan rata-rata 30,83 pelanggan per jam, yang sangat dekat dengan nilai teoritis 30 pelanggan per jam. Jam tersibuk terjadi pada jam ke-8 dengan 39 pelanggan. Hasil ini menunjukkan bahwa distribusi Poisson mampu memodelkan jumlah kedatangan pelanggan yang berfluktuasi secara acak di sekitar nilai rata-ratanya, sehingga sesuai digunakan untuk memodelkan proses kedatangan pelanggan pada interval waktu tetap.

#### Pertanyaan Analisis

5. Mengapa jumlah pelanggan tiap interval tidak sama walaupun $\lambda$ tetap?
6. Bagaimana pengaruh $\lambda$ terhadap rata-rata jumlah pelanggan?
7. Apa beda $\lambda$ sebagai parameter dengan mean hasil simulasi?
8. Jika $\lambda$ berubah menurut jam, bagaimana model sebaiknya dikembangkan?

#### 5. Mengapa jumlah pelanggan tiap interval tidak sama walaupun λ tetap?

Meskipun nilai λ tetap, distribusi Poisson bersifat acak sehingga jumlah pelanggan pada setiap interval waktu dapat berbeda-beda. Nilai λ hanya menyatakan rata-rata jumlah pelanggan yang diharapkan pada setiap interval, bukan jumlah yang pasti terjadi.

#### 6. Bagaimana pengaruh λ terhadap rata-rata jumlah pelanggan?

Nilai λ berbanding lurus dengan rata-rata jumlah pelanggan. Semakin besar λ, semakin besar rata-rata pelanggan yang datang pada setiap interval waktu. Sebaliknya, semakin kecil λ, semakin sedikit rata-rata pelanggan yang datang.

#### 7. Apa beda λ sebagai parameter dengan mean hasil simulasi?

λ merupakan parameter teoritis yang ditentukan sebelum simulasi dan menyatakan nilai harapan (expected value) distribusi Poisson. Sedangkan, mean hasil simulasi diperoleh dari data acak yang dihasilkan komputer. Nilai mean simulasi umumnya mendekati λ, tetapi tidak selalu sama persis karena dipengaruhi oleh variasi acak.

#### 8. Jika λ berubah menurut jam, bagaimana model sebaiknya dikembangkan?

Jika tingkat kedatangan pelanggan berbeda pada setiap jam, maka model sebaiknya menggunakan nilai λ yang berbeda untuk tiap interval waktu. Misalnya, jam sibuk dapat menggunakan λ yang lebih besar, sedangkan jam sepi menggunakan λ yang lebih kecil. Dengan demikian, simulasi akan lebih realistis karena mengikuti pola kedatangan pelanggan yang berubah sepanjang hari.

### Sesi C - Simulasi Distribusi Eksponensial

#### 5.1 Program

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
lam = 0.5
n = 1000

interarrival = np.random.exponential(scale=1/lam, size=n)

print("Mean teoritis:", 1/lam, "menit")
print("Mean simulasi:", interarrival.mean(), "menit")
print("Median       :", np.median(interarrival))
print("P90          :", np.percentile(interarrival, 90))

plt.hist(interarrival, bins=30, edgecolor="black")
plt.xlabel("Waktu antar kedatangan (menit)")
plt.ylabel("Frekuensi")
plt.title(f"Distribusi Eksponensial (lambda={lam})")
plt.show()
```

### 5.2 Replikasi Contoh Materi

Jika rata-rata pelanggan datang setiap 2 menit, $\lambda=0.5$. Peluang pelanggan datang dalam $\le3$ menit:

$P(T \le 3) = 1 - e^{-0.5(3)} \approx 0.7769$

Hasilnya sekitar 77,69%.

```python
import math

lam = 0.5
t = 3
prob = 1 - math.exp(-lam*t)

print(f"P(T<={t})={prob:.4f}")
print(f"Persentase : {prob*100:.2f}%")
```

### Sesi D - Proses Stokastik dan Markov Chain

#### 6.3 Program - Simulasi 30 Hari

```python
# 13. incl. Mulai dari Cerah, simulasi 30 hari.

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
states = ["Cerah", "Mendung", "Hujan"]

P = np.array([
    [0.6, 0.3, 0.1],
    [0.2, 0.5, 0.3],
    [0.1, 0.4, 0.5]
])

def simulate_markov(P, start_state=0, n_steps=30):
    current = start_state
    history = [current]
    for _ in range(n_steps-1):
        current = np.random.choice(len(P), p=P[current])
        history.append(current)
    return history

history = simulate_markov(P, start_state=0, n_steps=30)
print([states[i] for i in history])

counts = np.bincount(history, minlength=3)
plt.bar(states, counts)
plt.xlabel("State cuaca")
plt.ylabel("Jumlah hari")
plt.title("Markov Chain - 30 Hari")
plt.show()
```

### Analisa Singkat

Simulasi Markov Chain selama 30 hari menggambarkan perubahan kondisi cuaca berdasarkan probabilitas transisi antar-state. Dimulai dari kondisi Cerah, sistem berpindah ke state lain secara acak sesuai matriks transisi. Hasil simulasi menunjukkan bahwa state Cerah lebih sering muncul karena memiliki probabilitas bertahan terbesar (60%), sedangkan state Mendung dan Hujan muncul dengan frekuensi yang lebih rendah. Selain itu, terdapat beberapa periode ketika suatu state bertahan selama beberapa hari berturut-turut, yang merupakan karakteristik alami dari proses Markov.

Distribusi yang digunakan adalah **distribusi Poisson**, karena distribusi ini digunakan untuk memodelkan jumlah kejadian yang terjadi dalam suatu interval tertentu. Pada simulasi ini digunakan parameter λ = 3, yang menggambarkan rata-rata 3 kendaraan per menit. Sebanyak 1.000 data dibangkitkan menggunakan distribusi Poisson.

Hasil simulasi menghasilkan mean sebesar **2,959**, variance sebesar **3,0033**, nilai minimum **0**, dan nilai maksimum **9**. Mean simulasi mendekati nilai parameter λ = 3, sedangkan variance juga berada sangat dekat dengan nilai tersebut.

Histogram menunjukkan bahwa frekuensi terbesar berada pada sekitar **2–3 kendaraan per menit**, kemudian menurun ketika jumlah kendaraan semakin besar. Distribusi memiliki bentuk yang cenderung miring ke kanan, dengan jumlah kendaraan yang tinggi muncul lebih jarang. Hasil tersebut menunjukkan bahwa data hasil simulasi memiliki karakteristik yang sesuai dengan distribusi Poisson dengan λ = 3.

Berdasarkan hasil tersebut, simulasi menggambarkan kondisi dengan rata-rata sekitar **3 kendaraan yang melewati gerbang setiap menit**, sementara jumlah kendaraan pada setiap menit dapat berubah-ubah.

Sementara itu, pada skenario Ramai dengan λ = 5, distribusi bergeser lebih jauh ke kanan dengan frekuensi terbesar berada di sekitar 4–6 kendaraan per menit.

Hasil tersebut menunjukkan bahwa perubahan parameter λ memengaruhi jumlah kejadian yang dihasilkan oleh proses Poisson. ### Tugas 3 – Markov Chain Challenge

Pada simulasi ini digunakan tiga state untuk menggambarkan kondisi mesin, yaitu **Beroperasi, Pemeliharaan, dan Rusak**. Matriks probabilitas transisi yang digunakan memiliki jumlah setiap baris sebesar 1 dan seluruh nilai probabilitas berada pada rentang 0 hingga 1, sehingga matriks dapat digunakan dalam simulasi Markov Chain.

Simulasi dilakukan sebanyak 100 langkah dengan tiga state awal yang berbeda. Hasil simulasi menunjukkan bahwa dari state awal **Beroperasi**, mesin berada pada kondisi Beroperasi sebanyak 72 kali, Pemeliharaan 20 kali, dan Rusak 8 kali. Dari state awal **Pemeliharaan**, diperoleh 70 kali Beroperasi, 23 kali Pemeliharaan, dan 7 kali Rusak. Sementara itu, dari state awal **Rusak**, diperoleh 81 kali Beroperasi, 14 kali Pemeliharaan, dan 5 kali Rusak.

Pada ketiga simulasi, state **Beroperasi** memiliki frekuensi paling tinggi. Hal ini sesuai dengan matriks transisi karena probabilitas mesin untuk berada pada atau kembali ke kondisi Beroperasi cukup besar. Dari Pemeliharaan terdapat probabilitas 0,80 untuk berpindah ke Beroperasi, sedangkan dari Rusak terdapat probabilitas 0,60 untuk berpindah ke Beroperasi. Selain itu, ketika mesin berada dalam kondisi Beroperasi, probabilitas untuk tetap berada pada kondisi tersebut adalah 0,70.

Perbedaan state awal menyebabkan perbedaan frekuensi pada hasil simulasi, tetapi pola umumnya tetap menunjukkan bahwa kondisi Beroperasi paling sering muncul. Hasil tersebut juga menunjukkan bahwa state awal tidak sepenuhnya menentukan kondisi mesin sepanjang simulasi karena setiap state dapat berpindah ke state lain berdasarkan probabilitas transisi.

Dengan demikian, simulasi Markov Chain dapat digunakan untuk menggambarkan perubahan kondisi mesin dari satu langkah ke langkah berikutnya. Perubahan probabilitas transisi dapat digunakan untuk merepresentasikan kondisi sistem yang berbeda dan melihat bagaimana pola kondisi mesin berubah selama proses simulasi.
